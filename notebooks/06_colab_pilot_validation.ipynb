```python
import json, glob
records = {}
for path in glob.glob("outputs/metrics/pilot__pythia-160m_*.json"):
    with open(path) as f:
        rec = json.load(f)
    records[rec.get("compression_method")] = rec

print(f"{'arm':<14}{'perplexity':>12}{'retention %':>14}")
for method in ["dense", "pruning", "quantisation", "sequential", "joint"]:
    rec = records.get(method)
    if rec is None:
        print(f"{method:<14}  MISSING RECORD")
        continue
    ppl = rec["quality"]["perplexity"]["perplexity"]
    retention = rec["quality"].get("retention", {}).get("perplexity_retention")
    print(f"{method:<14}{ppl:>12.3f}{retention:>14.2f}" if retention is not None
          else f"{method:<14}{ppl:>12.3f}{'MISSING':>14}")
```

```python
joint_rec, seq_rec = records["joint"], records["sequential"]
stats = joint_rec["compression"]["statistics"]
print("match_sequential_budget:", stats.get("match_sequential_budget"))
print("joint optimiser_steps:", stats.get("optimiser_steps"))
print("joint total_optimiser_steps:", joint_rec["compression"]["total_optimiser_steps"])
print("sequential total_optimiser_steps:", seq_rec["compression"]["total_optimiser_steps"])
```

```python
import subprocess
commit_hash = subprocess.check_output(["git", "log", "--oneline", "-1"]).decode().strip()

dense_ppl = records["dense"]["quality"]["perplexity"]["perplexity"]
prune_ppl = records["pruning"]["quality"]["perplexity"]["perplexity"]
quant_ppl = records["quantisation"]["quality"]["perplexity"]["perplexity"]
seq_ppl = records["sequential"]["quality"]["perplexity"]["perplexity"]
joint_ppl = records["joint"]["quality"]["perplexity"]["perplexity"]

seq_ret = records["sequential"]["quality"].get("retention", {}).get("perplexity_retention", 0.0)
joint_ret = records["joint"]["quality"].get("retention", {}).get("perplexity_retention", 0.0)

diff_ret = joint_ret - seq_ret
joint_wins = diff_ret > 0
match_budget = stats.get('match_sequential_budget')

print("=" * 65)
print("FINAL COMPRESSION VERDICT")
print("=" * 65)
print(f"Commit: {commit_hash}")
print(f"Dense PPL:        {dense_ppl:.3f} | Retention: 100.00%")
print(f"Pruning PPL:      {prune_ppl:.3f}")
print(f"Quantisation PPL: {quant_ppl:.3f}")
print(f"Sequential PPL:   {seq_ppl:.3f} | Retention: {seq_ret:.2f}%")
print(f"Joint PPL:        {joint_ppl:.3f} | Retention: {joint_ret:.2f}%")
print("-" * 65)
if joint_wins:
    print(f"Verdict: Joint beat Sequential by +{diff_ret:.2f}% retention.")
else:
    print(f"Verdict: Sequential beat Joint by +{-diff_ret:.2f}% retention.")
print(f"match_sequential_budget: {match_budget}")
print("-" * 65)
print("IMPORTANT NOTICE:")
print("configs/experiments/pilot.yaml is pipeline validation only ")
print("(1 seed, 64 eval samples, 60 steps per file header) -- NOT a citable result.")
print("=" * 65)
```
