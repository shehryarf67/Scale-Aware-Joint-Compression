```python
import os

# Read the original file content
with open('src/scale_aware_compression/compression/activations.py', 'r') as f:
    lines = f.readlines()

# Filter out the broken append if it exists (lines starting with 'from __future__' at the end)
clean_lines = []
for line in lines:
    if line.strip() == 'from __future__ import annotations' and len(clean_lines) > 50:
        continue
    clean_lines.append(line)

activations_func = r'''
def capture_activations(model, dataloader, adapter, block_idx, group=None, inputs=None, device="cuda"):
    """Utility to capture activations for a specific block or group."""
    import torch
    from scale_aware_compression.compression.activations import LinearActivationCapture

    # Identify modules to capture
    named_modules = list(group) if group is not None else list(model.named_modules())
    target_modules = [(n, m) for n, m in named_modules if adapter.is_targeted(m)]

    captures = {name: LinearActivationCapture(mod, device=device) for name, mod in target_modules}

    with torch.no_grad():
        for cap in captures.values():
            cap.__enter__()

        if inputs is not None:
            # Advance model using cached inputs if provided by adapter
            _ = adapter.advance_activations(None, inputs, device=device)
        else:
            batch = next(iter(dataloader))
            if isinstance(batch, dict):
                model(**{k: v.to(device) for k, v in batch.items()})
            else:
                model(batch.to(device))

        for cap in captures.values():
            cap.__exit__()

    return {n: c.statistics for n, c in captures.items()}
'''

with open('src/scale_aware_compression/compression/activations.py', 'w') as f:
    f.writelines(clean_lines)
    f.write(activations_func)

print('Successfully corrected capture_activations in activations.py.')
```

```python
import os
import json
import pandas as pd

metrics_dir = 'outputs/metrics'
results = []

if os.path.exists(metrics_dir):
    files = [f for f in os.listdir(metrics_dir) if f.endswith('.json')]
    for filename in files:
        with open(os.path.join(metrics_dir, filename), 'r') as f:
            try:
                data = json.load(f)
                quality = data.get('quality', {})
                ppl_obj = quality.get('perplexity', {})
                retention_obj = quality.get('retention', {})

                # The metrics use 'perplexity' inside the perplexity object
                # and 'perplexity_retention' inside the retention object
                results.append({
                    'method': data.get('compression_method', 'unknown'),
                    'budget': data.get('budget_label', 'unknown'),
                    'perplexity': ppl_obj.get('perplexity'),
                    'retention': retention_obj.get('perplexity_retention')
                })
            except Exception as e:
                print(f"Error parsing {filename}: {e}")

df = pd.DataFrame(results)
print("\n--- Task 2: Pythia-1B Results Summary ---")
if not df.empty:
    summary = df.sort_values(['budget', 'perplexity'])
    print(summary.to_string(index=False))

    for budget in df['budget'].unique():
        b_df = df[df['budget'] == budget]
        # Exclude dense from 'best sequential' selection
        sequential_df = b_df[b_df['method'] != 'dense'].dropna(subset=['perplexity'])
        if not sequential_df.empty:
            best = sequential_df.loc[sequential_df['perplexity'].idxmin()]
            print(f"\nWinning Arm for {budget}: {best['method']} (PPL: {best['perplexity']:.4f})")
else:
    print("No results found in outputs/metrics.")
```

```python
import os
import json
import pandas as pd

metrics_dir = 'outputs/metrics'
results = []

if os.path.exists(metrics_dir):
    for filename in os.listdir(metrics_dir):
        if filename.endswith('.json'):
            with open(os.path.join(metrics_dir, filename), 'r') as f:
                try:
                    data = json.load(f)
                    quality = data.get('quality', {})
                    ppl_obj = quality.get('perplexity', {})
                    retention_obj = quality.get('retention', {})

                    # Updated to use the correct keys discovered in the actual metrics files
                    results.append({
                        'method': data.get('compression_method', 'unknown'),
                        'budget': data.get('budget_label', 'unknown'),
                        'perplexity': ppl_obj.get('perplexity'),
                        'retention': retention_obj.get('perplexity_retention')
                    })
                except Exception as e:
                    print(f"Error parsing {filename}: {e}")

df = pd.DataFrame(results)
print("--- Task 2: Pythia-1B Results Summary ---")
if not df.empty:
    print(df.sort_values(['budget', 'perplexity']).to_string(index=False))

    # Identify winners (excluding dense from the sequential/joint comparison)
    for budget in df['budget'].unique():
        b_df = df[df['budget'] == budget]
        # Exclude dense from 'best' selection to find the winning compression arm
        comp_df = b_df[b_df['method'] != 'dense'].dropna(subset=['perplexity'])
        if not comp_df.empty:
            best = comp_df.loc[comp_df['perplexity'].idxmin()]
            print(f"\nWinning Arm for {budget}: {best['method']} (PPL: {best['perplexity']:.4f})")
else:
    print("No results found.")
```

```python
import os

# Inspect the runner and evaluation code to find CPU overrides
paths_to_check = [
    'scripts/run_scale_sweep.py',
    'src/scale_aware_compression/experiments/runner.py',
    'src/scale_aware_compression/evaluation/common.py'
]

for path in paths_to_check:
    if os.path.exists(path):
        with open(path, 'r') as f:
            content = f.read()
            if 'device' in content or 'cpu' in content.lower():
                print(f'--- {path} ---')
                # Look for lines that look like they might be forcing CPU
                lines = content.split('\n')
                for i, line in enumerate(lines):
                    if 'device =' in line or 'device=' in line or 'cpu' in line.lower():
                        print(f'{i+1}: {line.strip()}')
                print('\n')
```

```python
import os

runner_path = 'src/scale_aware_compression/experiments/runner.py'
if os.path.exists(runner_path):
    with open(runner_path, 'r') as f:
        content = f.read()

    # Replace the forced CPU move with a move to the configured device
    # Old: model.to("cpu")
    # New: model.to(config.evaluation.device.value)
    patched_content = content.replace('model.to("cpu")', 'model.to(config.evaluation.device.value)')
    patched_content = patched_content.replace('"evaluate quality (CPU)"', 'f"evaluate quality ({config.evaluation.device.value})"')

    with open(runner_path, 'w') as f:
        f.write(patched_content)

print("Successfully patched runner.py to respect evaluation.device.")
```
